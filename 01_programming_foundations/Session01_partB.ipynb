**GC content**

$\dfrac{1}{L}\sum_{i=1}^{L} \mathbb{1}[\,b_i \in \{G, C\}\,]$

It calculates the fraction of a sequence that is G or C

```python
#in more lines

#initialising the necessary variables
sequence='AATTAGACCCATTACGC'
L=len(sequence)
GC=0

#counting the number of Cs and Gs
for base in sequence:
  if base=='G'or base=='C':
    GC+=1

#calculating the gc_content with the according formula
gc_content= GC/L

#print the result
print(f"\n C content is:{ gc_content} \n Percentage: {gc_content*100}")
```

```python
#in a single line
sequence='AATTAGACCCATTACGC'


#calculating the gc_content with the according formula
gc_content=1/len(sequence)*len([base for base in sequence if base=='C' or base=='G'])

#print the result
print(f"\n C content is:{ gc_content} \n Percentage: {gc_content*100}")
```
